# Stopword Removal

```python
# 3.1 Import Library Stopword
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords

# Stopword bawaan NLTK Bahasa Indonesia
stop_words = set(stopwords.words('indonesian'))

# Tambahan stopword tidak baku 
additional_stopwords = {
    'yg', 'utk', 'tdk', 'lg', 'aja', 'dll', 
    'nya', 'nih', 'sih', 'dong', 'kok'
}

# Gabungkan stopword bawaan + tambahan
stop_words = stop_words.union(additional_stopwords)

print("Total stopwords:", len(stop_words))
```

```python
# 3.2 Load Dataset Setelah Tokenization
df = pd.read_csv('datapreprocessing/data_after_tokenization.csv')

# Karena teks_token tersimpan sebagai string list, kita ubah kembali ke list
import ast
df['teks_token'] = df['teks_token'].apply(ast.literal_eval)

print("Dataset berhasil dimuat.")
```

```python
# 3.3 Proses Stopword Removal

df['teks_stopword'] = df['teks_token'].apply(
    lambda tokens: [word for word in tokens if word not in stop_words]
)

print("Stopword removal selesai.")
df[['teks_token', 'teks_stopword']].head()
```

```python
# 3.4 Simpan Checkpoint Setelah Stopword Removal

df.to_csv(
    os.path.join('datapreprocessing', 'data_after_stopword_removal.csv'),
    index=False
)

print("data_after_stopword_removal.csv berhasil disimpan.")
```

# Stemming

```python
# 4.1 Install dan Import Library Sastrawi 
import sys
!{sys.executable} -m pip install Sastrawi
```

```python
# 4.2 Load Dataset Setelah Stopword Removal
df = pd.read_csv('datapreprocessing/data_after_stopword_removal.csv')

df['teks_stopword'] = df['teks_stopword'].apply(ast.literal_eval)

print("Dataset berhasil dimuat.")
```

```python
# 4.3 Inisialisasi Stemmer
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

factory = StemmerFactory()
stemmer = factory.create_stemmer()
```

```python
# 4.4 Proses Stemming

def stemming_process(tokens):
    return [stemmer.stem(word) for word in tokens]

df['teks_stemming'] = df['teks_stopword'].apply(stemming_process)

print("Stemming selesai.")
df[['teks_stopword', 'teks_stemming']].head()
```

```python
# 4.5 Menggabungkan Kembali Token Menjadi Kalimat

df['teks_final'] = df['teks_stemming'].apply(lambda tokens: ' '.join(tokens))

df[['teks_stemming', 'teks_final']].head()
```

```python
# 4.6 Simpan Dataset Hasil Preprocessing

df.to_csv(
    os.path.join('datapreprocessing', 'data_preprocessing_final.csv'),
    index=False
)

print("data_preprocessing_final.csv berhasil disimpan.")
```
